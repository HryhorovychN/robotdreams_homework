```python
import torch
import cv2
import numpy as np
import json
from pathlib import Path

# Path to the trained model
model_path = r'C:\Users\user\Documents\GitHub\robotdreams_homework\project\yolov5\runs\train\exp2\weights\best.pt'

# Load the trained YOLOv5 model
model = torch.hub.load('ultralytics/yolov5', 'custom', path=model_path, force_reload=True)
model.eval()

# Define a color map for different classes
color_map = {
    0: (255, 0, 0),   # Blue for class 0
    1: (0, 255, 0),   # Green for class 1
    2: (0, 0, 255),   # Red for class 2
    3: (255, 255, 0), # Cyan for class 3
    # Add more classes if needed
}

# Function to process video stream from webcam
def detect_objects(min_confidence=0.5):
    cap = cv2.VideoCapture(0)  # Use the first webcam
    
    # Initialize an empty dictionary for object counts
    object_counts = {}

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Perform detection
        results = model(frame)

        # Get detection information and visualize
        for result in results.xyxy[0].cpu().numpy():
            x1, y1, x2, y2, confidence, class_idx = result

            # Check if confidence is above the threshold
            if confidence >= min_confidence:
                label = f'{model.names[int(class_idx)]} {confidence:.2f}'

                # Get color for the current class
                color = color_map.get(int(class_idx), (255, 255, 255))  # Default to white if class index not in color_map

                # Draw bounding box and label
                cv2.rectangle(frame, (int(x1), int(y1)), (int(x2), int(y2)), color, 2)
                cv2.putText(frame, label, (int(x1), int(y1) - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 2)

                # Increment object count
                class_name = model.names[int(class_idx)]
                if class_name not in object_counts:
                    object_counts[class_name] = 0
                object_counts[class_name] += 1

        # Display video stream
        cv2.imshow('Object Detection', frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    cap.release()
    cv2.destroyAllWindows()

    return object_counts

# Function to save object counts to a JSON file
def save_object_counts(object_counts, output_path):
    with open(output_path, 'w') as f:
        json.dump(object_counts, f, indent=4)

# Path to save the object counts as a JSON file
output_folder = r'C:\Users\user\Documents\GitHub\robotdreams_homework\project\test_results'
Path(output_folder).mkdir(parents=True, exist_ok=True)
object_counts_output_path = Path(output_folder) / 'object_counts.json'

# Run detection on video stream
object_counts = detect_objects()

# Print and save object counts
for obj_name, count in object_counts.items():
    print(f'{obj_name}: {count}')

# Save object counts to a JSON file
save_object_counts(object_counts, object_counts_output_path)
```
